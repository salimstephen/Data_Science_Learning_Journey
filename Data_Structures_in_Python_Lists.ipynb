# Data Structures in Python: Lists

## Overview
This notebook explores the **list** data structure in Python. Lists are ordered and mutable collections that allow us to store, modify, and organize multiple values efficiently.

The exercises below demonstrate how to create lists, manipulate their contents, work with nested lists, and sort list elements.


## Learning Objectives
By the end of this exercise, you will be able to:

- Create lists in Python
- Manipulate and slice lists
- Append and combine lists
- Work with nested lists
- Sort list elements alphabetically


## Exercise 1: Creating and Manipulating Lists

In this exercise, we work with lists to represent trees planted in a plantation and practice common list operations.


```python
# Create a list containing integers from 1 to 5
first_trees = [1, 2, 3, 4, 5]

first_trees
```

### Alternative Way to Create a List

Instead of specifying each element individually, Python provides the `range()` function for generating sequences of numbers.


```python
# Create the same list using range()
first_trees = list(range(1, 6))

first_trees
```

### Creating a List with a Specific Range and Increment

We now create a list of trees planted within 15 cm of an irrigation device.  
The list starts at 20, ends at 140, and increases in steps of 4.


```python
tree_list = list(range(20, 141, 4))

tree_list
```

### Appending Elements to a List

Lists are mutable, meaning we can add new elements after creation.  
We append the value `400` to represent an older irrigated tree.


```python
tree_list.append(400)

tree_list
```

### Combining Lists

We install irrigation for additional older trees and combine the existing list with another list using the `+` operator.


```python
old_trees = [500, 600, 700]

combined_trees = tree_list + old_trees

len(combined_trees), combined_trees
```

### Working with Nested Lists

A nested list is a list that contains other lists as its elements.  
Below, we group conifers and cycads into a broader family called gymnosperms.


```python
conifers = ["Pine", "Fir", "Juniper"]
cycads = ["Cycas", "Sago palm", "Zamia"]

gymnosperms = [conifers, cycads]

gymnosperms
```

### Accessing Elements in Nested Lists

Using indexing, we extract specific plants from within a nested list structure.


```python
selected_plants = [gymnosperms[1][0], gymnosperms[1][1]]

selected_plants
```

### Combining and Sorting List Elements

We combine the conifers and cycads into a single list and sort the plants alphabetically.


```python
ordered_plants = conifers + cycads
ordered_plants.sort()

ordered_plants
```
