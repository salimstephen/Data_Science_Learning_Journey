# Data Structures in Python: Dictionaries

## Overview
This notebook explores the **dictionary** data structure in Python. Dictionaries store data as **key–value pairs** and are widely used for representing structured and labeled information.

The exercises below focus on creating dictionaries, modifying their contents, extracting data, and working with nested dictionaries.


## Learning Objectives
By the end of this exercise, you will be able to:

- Create dictionaries in Python
- Access values using dictionary keys
- Add, update, and remove key–value pairs
- Extract keys, values, and items from dictionaries
- Work with nested dictionaries


## Exercise 1: Creating and Manipulating Dictionaries

In this exercise, we create a dictionary representing information about a tree and practice common dictionary operations.


```python
# Create a dictionary with tree information
tree = {
    "tree_type": "Pepper-bark tree",
    "age": 15,
    "family": "Canellaceae"
}

tree
```

### Accessing Values by Key

Dictionary values can be accessed using their corresponding keys.


```python
# Print the value associated with the key 'age'
tree["age"]
```

### Adding New Information

Dictionaries are mutable, allowing us to add new key–value pairs after creation.


```python
# Add the kingdom information
tree["kingdom"] = "Plantae"

tree
```

### Removing Information from a Dictionary

We can remove a key–value pair using the `del` keyword.


```python
# Remove the 'family' key-value pair
del tree["family"]

tree
```

### Updating Dictionary Values

Existing dictionary values can be updated by reassigning a value to an existing key.


```python
# Update the tree type
tree["tree_type"] = "Wild Cinnamon"

tree
```

### Extracting Keys and Values

The `keys()` and `values()` methods can be used to extract dictionary data into lists.


```python
# Extract keys and values into separate lists
key_list = list(tree.keys())
value_list = list(tree.values())

key_list, value_list
```

### Extracting Key-Value Pairs as Tuples

The `items()` method returns key–value pairs as tuples, which can be stored in a list.


```python
# Extract key-value pairs into a list of tuples
item_list = list(tree.items())

item_list
```

## Exercise 2: Working with Nested Dictionaries

In this exercise, we update a nested dictionary containing information about African tree species by adding additional attributes.


```python
# Basic nested dictionary
african_trees = {
    "Baobab": {
        "common_name": "Baobab",
        "scientific_name": "Adansonia",
    },
    "Acacia": {
        "common_name": "Acacia",
        "scientific_name": "Acacia",
    }
}

african_trees
```

### Adding New Information to Nested Dictionaries

We add new keys (`country` and `average_height`) to each nested dictionary using key-based access.


```python
# Add additional information for Baobab
african_trees["Baobab"]["country"] = "Zimbabwe"
african_trees["Baobab"]["average_height"] = 25

# Add additional information for Acacia
african_trees["Acacia"]["country"] = "South Africa"
african_trees["Acacia"]["average_height"] = 15

african_trees
```
