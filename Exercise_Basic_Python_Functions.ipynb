## Exercise 1: Total Carbon Emission Calculation

### Problem Description
We are given a list representing deforested land (in hectares) over five years.
The goal is to calculate the **total carbon emissions** produced over this period.

### Formula
Total Carbon Emissions =  
**(Total deforested land) × (Carbon emission factor)**

### Key Concepts Used
- Built-in function: `sum()`
- Variable assignment
- Basic arithmetic operations


```python
# Deforested land for each year (in hectares)
deforested_land = [10, 15, 8, 12, 20]

# Carbon emission factor (tons of CO2 per hectare)
carbon_emission_factor = 40

# Calculate total deforested land using the built-in sum function
total_deforested_land = sum(deforested_land)

# Calculate total carbon emissions
total_carbon_emission = total_deforested_land * carbon_emission_factor

# Print results
print("Total deforested land (hectares):", total_deforested_land)
print("Total carbon emission (tons of CO2):", total_carbon_emission)
```

## Exercise 2: Acres to Hectares Conversion

### Problem Description
Create a user-defined function that converts forest area from **acres to hectares**.

### Conversion Rate
1 acre = **0.404686 hectares**

### Requirements
- The function must be named `acres_to_hectares`
- It should accept one parameter
- It must return the converted value
- Include a proper docstring


```python
def acres_to_hectares(acres):
    """
    Converts land area from acres to hectares.

    Parameters:
    acres (float): Area of land in acres

    Returns:
    float: Area of land in hectares
    """
    conversion_rate = 0.404686
    return acres * conversion_rate
```

```python
# Convert 50 acres to hectares
forest_area_hectares = acres_to_hectares(50)

# Print result
print("Forest area in hectares:", forest_area_hectares)
```

## Exercise 3: Average Deforested Land

### Problem Description
Using historical deforestation data, calculate the **average deforested land per year**.

### Key Concepts Used
- Built-in functions: `sum()` and `len()`
- Basic arithmetic


```python
# Deforested land for each year (in hectares)
deforested_land = [10, 15, 8, 12, 20]

# Calculate average deforested land
average_deforested_land = sum(deforested_land) / len(deforested_land)

# Print result
print("Average deforested land per year (hectares):", average_deforested_land)
```
