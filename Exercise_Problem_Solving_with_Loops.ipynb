## Exercise 1: Ocean Cleanup Analysis

### Problem Description
An ocean cleanup campaign collects waste every day for one week.
Each day's collected waste (in kilograms) is stored in a list.

Your task is to calculate the **total amount of waste collected over the week**.

### Key Concepts Used
- `for` loop
- Accumulator variable
- List iteration


```python
# List containing daily waste collection (in kilograms)
weekly_collection = [340, 220, 455, 300, 385, 520, 410]

# Initialize total waste collected
total_waste_collected = 0

# Loop through each day's collection and add it to the total
for daily_collection in weekly_collection:
    total_waste_collected += daily_collection

# Display the result
print(f"Total waste collected over the week: {total_waste_collected} kg")
```

## Exercise 2: Endangered Species Monitoring

### Problem Description
A wildlife monitoring system records animal sightings.
The program should scan through the sightings and **stop immediately** when an endangered species is detected.

### Key Concepts Used
- `for` loop
- Conditional statements (`if`)
- Loop control using `break`


```python
# List of endangered species
endangered_species = ['Tiger', 'Blue Whale', 'Mountain Gorilla']

# Recorded sightings
sightings = ['Deer', 'Tiger', 'Blue Whale', 'Mountain Gorilla', 'Elephant']

# Loop through the sightings
for species in sightings:
    if species in endangered_species:
        print(f"Endangered species spotted: {species}")
        break
```

## Exercise 3: Forest Growth Monitoring

### Problem Description
A forest starts with an initial area and grows by a fixed percentage each year.
The program should simulate the forest's growth **until it exceeds a target area**.

### Key Concepts Used
- `while` loop
- Compound growth
- Loop counters
- Formatted output


```python
# Initial forest area in square kilometres
forest_area = 1000

# Annual growth rate (5%)
growth_rate = 1.05

# Year counter
year = 0

# Simulate forest growth until it reaches the target size
while forest_area < 1500:
    year += 1
    forest_area *= growth_rate
    print(f"Forest area after {year} years: {forest_area:.2f} sq km")
```
