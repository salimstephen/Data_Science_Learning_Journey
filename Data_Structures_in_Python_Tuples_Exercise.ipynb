- Unpack tuple values into variables

## Exercise 1: Creating and Working with Tuples

In this exercise, we create a tuple, explore its properties, and perform basic operations such as indexing, slicing, and checking immutability.


```python
# assign value to the tree_tuple
tree_tuple = (8, 'apple', 'Granny Smith')

# see what it looks like printed
print(tree_tuple)
```

```python
# confirm that we have created a tuple using the `type()` function
type(tree_tuple)
```

### Finding the Length of a Tuple

The `len()` function is used to determine the number of elements contained in a tuple.


```python
len(tree_tuple)
```

### Accessing Tuple Elements

Tuple elements can be accessed using indexing. Indexing in Python starts from `0`.


```python
# Access the second element
fruit = tree_tuple[1]
print(fruit)
```

### Tuple Immutability

Tuples are immutable, meaning their elements cannot be changed after creation.  
Attempting to modify a tuple will result in an error.


```python
# This will raise a TypeError
tree_tuple[1] = 'pear'
```

### Slicing a Tuple

Slicing allows us to extract a subset of elements from a tuple and store them in a new tuple.


```python
# Create a new tuple containing the last two elements
text_only = tree_tuple[1:]
text_only
```

```python
# Another solution makes use of negative indexing, leaving the second element empty
text_only = tree_tuple[-2:]
print(text_only)
```

## Exercise 2: Tuple Operations

This exercise focuses on tuple concatenation, tuple methods, and tuple unpacking.


```python
# Create summer and winter fruit tuples
fruit_summer = ('peach', 'apricot', 'plum')
fruit_winter = ('lemon', 'orange', 'grapefruit')

# Print both tuples
print(fruit_summer, fruit_winter)
```

### Concatenating Tuples

Tuples can be combined using the `+` operator to create a new tuple.


```python
# Combine both tuples
all_fruit = fruit_summer + fruit_winter
print(all_fruit)
```

### Using Tuple Methods

The `count()` method is used to determine how many times a specific element appears in a tuple.


```python
all_fruit.count('peach')
```

### Tuple Unpacking

Tuple unpacking allows us to assign each element of a tuple to a separate variable in a single statement.


```python
# Given tuple
combined_info = ('yellow', 5, 'lemon')

# Unpack the tuple
colour, age, tree_type = combined_info

print(colour)
print(age)
print(tree_type)
```
