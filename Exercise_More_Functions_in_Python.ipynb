## Exercise 1: Analyse Overall Carbon Impact

### Problem Description
We are tasked with analysing the overall carbon impact based on:
- Deforestation emissions
- Tree planting offsets

### Logic
1. Calculate total carbon emissions:
   - Sum of (deforested area × emission factor)
2. Calculate total carbon offsets:
   - Sum of (planted trees × offset per tree)
3. Compare totals and categorise the impact:
   - **Positive** → total offset ≥ total emission
   - **Neutral** → total offset == total emission
   - **Negative** → total offset < total emission


```python
def analyse_carbon_impact(deforested_areas, planted_trees, carbon_emission_factor, carbon_offset_per_tree):
    """
    Calculate and categorise the overall carbon impact, considering deforestation,
    tree planting, and carbon offset.

    Returns:
    - str: 'Positive', 'Neutral', or 'Negative'
    """
    total_emission = sum(area * carbon_emission_factor for area in deforested_areas)
    total_offset = sum(trees * carbon_offset_per_tree for trees in planted_trees)

    if total_offset > total_emission:
        return "Positive"
    elif total_offset == total_emission:
        return "Neutral"
    else:
        return "Negative"
```

```python
# Given data
deforested_areas = [10, 15, 8, 12, 20]
planted_trees = [30, 40, 25, 35, 50]
carbon_emission_factor = 30
carbon_offset_per_tree = 5

impact_category = analyse_carbon_impact(
    deforested_areas,
    planted_trees,
    carbon_emission_factor,
    carbon_offset_per_tree
)

print("Overall carbon impact category:", impact_category)
```

## Exercise 2: Project Future Tree Planting

### Problem Description
This exercise projects future tree planting numbers based on:
- Initial planting count
- Annual growth rate
- Number of projection years

### Approach
We use a **list comprehension** to calculate projections for each year using the formula:

initial_planted_trees × (1 + annual_growth_rate) ** year


```python
def project_future_tree_planting(initial_planted_trees, annual_growth_rate, projection_years):
    """
    Project the future number of planted trees for each year.

    Returns:
    - list: Projected number of trees for each year
    """
    projections = [
        int(initial_planted_trees * (1 + annual_growth_rate) ** year)
        for year in range(1, projection_years + 1)
    ]

    return projections
```

```python
initial_planted_trees = 100
annual_growth_rate = 0.05
projection_years = 10

projected_trees = project_future_tree_planting(
    initial_planted_trees,
    annual_growth_rate,
    projection_years
)

print("Projected tree planting over years:")
print(projected_trees)
```

## Exercise 3: Calculate Regional Carbon Offset

### Problem Description
We calculate a region’s carbon offset dynamically using multiple contributing factors.

### Key Concept
- `**kwargs` allows us to accept a variable number of named parameters.
- The offset is calculated as the **average** of all provided factor values.

### Output
The function returns a tuple:
(region_name, calculated_carbon_offset)


```python
def calculate_region_offset_with_kwargs(region_name, **factors):
    """
    Calculate dynamic carbon offset for a region based on provided factors.

    Returns:
    - tuple: (region_name, average carbon offset)
    """
    average_offset = sum(factors.values()) / len(factors)

    return region_name, average_offset
```

```python
region, offset = calculate_region_offset_with_kwargs(
    "Amazon",
    deforested_area=30,
    planted_trees=40,
    carbon_emission_factor=10
)

print(f"Region: {region}")
print(f"Calculated carbon offset: {offset}")
```
