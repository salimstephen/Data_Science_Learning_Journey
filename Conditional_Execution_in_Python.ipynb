## Conditional Statements: `if`, `elif`, and `else`

Conditional statements allow a program to execute specific blocks of code based on whether a given condition evaluates to `True` or `False`. They form the basis of decision-making in Python programs and control the flow of execution.


```python
# Check if the deforestation rate exceeds 20% and print a warning if it does
deforestation_rate = 0.25  # 25%

if deforestation_rate > 0.2:
    print("Critical deforestation level reached!")
```

## Nested `if` Statements

Nested `if` statements allow us to perform additional checks within an existing conditional block. This is useful when multiple conditions must be satisfied before an action is taken.


```python
# Check if deforestation exceeds 15% in a protected area
deforestation_rate = 0.18
protected_area = True

if deforestation_rate > 0.15:
    if protected_area:
        print("Urgent action needed in protected area due to high deforestation rate!")
```

## `if-else` Statements

The `if-else` structure executes one block of code when a condition is `True` and another when it is `False`. This ensures that exactly one outcome is executed.


```python
# Display deforestation alert status
deforestation_alert = True

if deforestation_alert:
    print("Alert: Deforestation detected!")
else:
    print("No deforestation alerts.")
```

## `elif` Statements

The `elif` keyword (short for "else if") allows us to evaluate multiple conditions sequentially. It is useful when we need to categorise values into more than two possible outcomes.


```python
# Classify deforestation levels
deforestation_rate = 0.10

if deforestation_rate > 0.2:
    print("Critical level")
elif deforestation_rate > 0.1:
    print("High level")
else:
    print("Normal level")
```

## Chained Conditional Statements

Chained conditionals use multiple `elif` statements to handle scenarios where several conditions are possible, but only one should be executed. The program evaluates conditions from top to bottom and stops at the first `True` condition.


```python
# Determine actions based on deforestation severity
deforestation_rate = 0.12

if deforestation_rate > 0.3:
    print("Implement emergency measures")
elif deforestation_rate > 0.2:
    print("Increase surveillance and conservation efforts")
elif deforestation_rate > 0.1:
    print("Monitor and plan reforestation")
else:
    print("Deforestation rate is within safe limits")
```

### Important Note on Overlapping Conditions

When multiple conditions could evaluate to `True`, Python executes **only the first matching condition**. Therefore, conditions should be ordered carefully to avoid logical errors.


```python
deforestation_rate = 0.1

if deforestation_rate > 0.3:
    print("Implement emergency measures")
elif deforestation_rate > 0.2:
    print("Increase surveillance and conservation efforts")
elif deforestation_rate == 0.1:
    print("Deforestation has reached the maximum safe limit")
elif deforestation_rate >= 0.1:
    print("Monitor and plan reforestation")
else:
    print("Deforestation rate is within safe limits")
```

## Conditional Keywords: `not` and `pass`

Python provides additional keywords to enhance conditional logic:

- `not`: Tests the negation of a condition.
- `pass`: Acts as a placeholder where a statement is syntactically required but no action is needed.


```python
# Check forest health
deforestation_detected = False

if not deforestation_detected:
    print("Forest is healthy!")
else:
    print("Deforestation detected!")
```

### The `pass` Statement

The `pass` statement prevents syntax errors when a conditional block is intentionally left empty. It is commonly used as a placeholder for future logic.


```python
deforestation_rate = 0.22

if deforestation_rate > 0.1:
    pass  # Placeholder for future alert logic
else:
    print("No immediate action required")
```

## Ternary Operator

The ternary operator allows conditional logic to be written in a single line. It is ideal for simple conditions and improves code readability when used appropriately.



```python
# Determine deforestation status
deforestation_rate = 0.07
status = "High" if deforestation_rate > 0.1 else "Normal"

print(f"Deforestation status: {status}")
```
