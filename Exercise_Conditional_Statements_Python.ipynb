## Exercise 1: Animal Risk Classification

Use conditional statements to categorise an animal based on its risk status.

### Categories:
- **Critically Endangered**: Less than 1,000 individuals  
- **Endangered**: 1,000 to 5,000 individuals  
- **Vulnerable**: 5,001 to 20,000 individuals  
- **Safe**: More than 20,000 individuals  

Assume there are **5,500 leopards** left in South Africa.


```python
individuals_in_wild = 5500

if individuals_in_wild < 1000:
    print("Critically Endangered")
elif individuals_in_wild <= 5000:
    print("Endangered")
elif individuals_in_wild <= 20000:
    print("Vulnerable")
else:
    print("Safe")
```

## Exercise 2: Water Scarcity Assessment

Determine whether a region is experiencing water scarcity based on:

- **Rainfall** (average annual rainfall in mm)
- **Population density** (people per square kilometre)

### Criteria:
- **Severe Scarcity**:
  - Rainfall < 500mm **and**
  - Population density > 500
- **Moderate Scarcity**:
  - Rainfall between 500mm and 1000mm (inclusive)
  - Population density between 200 and 500 (inclusive)
- **No Scarcity**: All other cases

Assume:
- Rainfall = 205mm  
- Population density = 448


```python
rainfall = 205
population_density = 448

if rainfall < 500 and population_density > 500:
    print("Severe Scarcity")
elif 500 <= rainfall <= 1000 and 200 <= population_density <= 500:
    print("Moderate Scarcity")
else:
    print("No Scarcity")
```

## Exercise 3: Recycling Classification

Categorise recyclable items based on their **material** and **size**.

### Possible Inputs:
- **Material**: paper, plastic, metal, glass
- **Size**: small, medium, large

### Recycling Categories:
- **Special Handling**: Large metal or glass items
- **Standard Recycling**:
  - All medium-sized items
  - Small metal or glass items
- **Composting**: Small paper items
- **Landfill**:
  - Small plastic items
  - All other cases


```python
material = "paper"
size = "small"

if size == "large" and material in ["metal", "glass"]:
    print("Special Handling")
elif size == "medium" or (size == "small" and material in ["metal", "glass"]):
    print("Standard Recycling")
elif size == "small" and material == "paper":
    print("Composting")
else:
    print("Landfill")
```

## Exercise 4: Renewable Energy Usage

Evaluate whether a household uses renewable energy.

- The variable `solar` stores whether the household uses solar energy (`True` or `False`).
- If the household **does not** use solar energy, print **"Needs Improvement"**.
- Otherwise, print **"Eco-friendly"**.

**Note:** This exercise must be solved using `if not`.


```python
solar = True

if not solar:
    print("Needs Improvement")
else:
    print("Eco-friendly")
```

## Exercise 5: Deforestation Alert

Issue an alert if deforestation exceeds a certain threshold.

- Define a variable `deforestation_rate` (a value between 0 and 1).
- If the rate is above **0.1**, print **"High Alert"**.
- If it is not, nothing should be printed.


```python
deforestation_rate = 0.15

if deforestation_rate > 0.1:
    print("High Alert")
```
