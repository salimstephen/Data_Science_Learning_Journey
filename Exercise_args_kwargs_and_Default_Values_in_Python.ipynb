## Exercise 1: Forest Area Calculation

### Problem Description
We want to calculate the **net forest area** covered by a group of trees while accounting for deforestation.

### Function Requirements
The function must:
- Be named `calculate_forest_area`
- Accept the following parameters:
  - `trees` (required)
  - `area_per_tree` (default value: 10 square metres)
  - `deforestation_factor` (default value: 0.2)
- Calculate the total forest area
- Subtract the deforested portion
- Return the remaining (net) forest area


```python
def calculate_forest_area(trees, area_per_tree=10, deforestation_factor=0.2):
    """
    Calculates the net forest area after accounting for deforestation.

    Parameters:
    trees (int): Number of trees in the forest
    area_per_tree (float): Area covered by one tree (default is 10 sq metres)
    deforestation_factor (float): Percentage of area deforested (default is 0.2)

    Returns:
    float: Net forest area after deforestation
    """
    total_area = trees * area_per_tree
    deforested_area = total_area * deforestation_factor
    net_forest_area = total_area - deforested_area

    return net_forest_area
```

```python
# Calculate forest area with custom values
forest_area = calculate_forest_area(
    trees=50,
    area_per_tree=5,
    deforestation_factor=0.3
)

print("Net forest area (sq metres):", forest_area)
```

## Exercise 2: Tree Planting Campaign Tracker

### Problem Description
This exercise tracks the growth of a forest over multiple tree planting campaigns.

### Key Concept
We use `*args` to allow the function to accept **any number of campaigns**, making the function flexible and reusable.

### Function Requirements
- Function name: `tree_planting_campaign`
- Parameters:
  - `initial_trees` (starting number of trees)
  - `*trees_per_campaign` (number of trees planted in each campaign)
- Return the final total number of trees


```python
def tree_planting_campaign(initial_trees, *trees_per_campaign):
    """
    Calculates the total number of trees after multiple planting campaigns.

    Parameters:
    initial_trees (int): Initial number of trees
    *trees_per_campaign (int): Trees added in each campaign

    Returns:
    int: Total number of trees after all campaigns
    """
    total_trees = initial_trees

    for trees in trees_per_campaign:
        total_trees += trees

    return total_trees
```

```python
# Track tree planting campaigns
final_tree_count = tree_planting_campaign(100, 20, 30, 50)

print("Total number of trees after campaigns:", final_tree_count)
```

## Exercise 3: Deforestation Impact Report

### Problem Description
We want to generate a clear report describing the impacts of deforestation.

### Key Concept
Using `**kwargs` allows us to pass a **variable number of named impact indicators**, making the function highly adaptable.

### Function Requirements
- Function name: `deforestation_impact_report`
- Accept `**impact_values`
- Print a labelled list of impact types and their values


```python
def deforestation_impact_report(**impact_values):
    """
    Prints a report of deforestation impacts and their values.

    Parameters:
    **impact_values: Keyword arguments representing impact types and values
    """
    print("Deforestation Impact Report")
    print("-" * 30)

    for impact, value in impact_values.items():
        print(f"{impact.replace('_', ' ').title()}: {value}")
```

```python
# Generate deforestation impact report
deforestation_impact_report(
    loss_of_biodiversity="30%",
    carbon_emission=500,
    soil_erosion=20
)
```
