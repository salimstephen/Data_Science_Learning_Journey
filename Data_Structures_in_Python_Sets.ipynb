# Data Structures in Python: Sets

## Overview
This notebook explores the **set** data structure in Python. Sets are unordered collections that store **unique elements** and do not support indexing.

Through practical exercises, we demonstrate how to create sets, compare them, and apply common set operations such as union, intersection, and difference.


## Learning Objectives
By the end of this exercise, you will be able to:

- Convert lists into sets
- Identify the core characteristics of sets
- Understand uniqueness and unordered behavior in sets
- Test membership within a set
- Apply set operations such as union, intersection, and difference
- Perform subset checks


## Exercise 1: Creating and Exploring Sets

In this exercise, we analyze a list of tree species and convert it into a set to identify unique tree types.


```python
# Creating a list with duplicate tree names
tree_list = [
    'Tamboti', 'Baobab', 'Umbrella thorn', 'Yellowwood', 'Fever Tree', 'Moringa',
    'Marula', 'Waterpear', 'Baobab', 'Fountain Tree', 'Papaya Tree',
    'Indian Laurel', 'Moringa'
]

# Convert list to set
tree_set = set(tree_list)

# Confirm the data type
type(tree_set)

# Print both variables
print("List:", tree_list)
print("Set:", tree_set)
```

### Comparing the Number of Elements

Since sets only store unique values, converting a list with duplicates into a set usually reduces the total number of elements.


```python
print("Number of elements in tree_list:", len(tree_list))
print("Number of elements in tree_set:", len(tree_set))
```

## Exercise 2: Set Operations

In this exercise, we compare tree species observed on two different days using set operations.


```python
# Trees observed on day 1
day_1_trees = {
    'Baobab', 'Buffalo Thorn', 'Bushwillow', 'Jackalberry',
    'Knob Thorn', 'Lala Palm', 'Marula Tree', 'Umbrella Thorn',
    'Weeping Boer Bean', 'Sausage Tree'
}

# Trees observed on day 2
day_2_trees = {
    'Mopane Tree', 'Bushwillow', 'Knob Thorn', 'Jackalberry',
    'Nara Plant', 'Wild Date Palm', 'Natal Mahogany',
    'Tamboti', 'Leadwood'
}

day_1_trees, day_2_trees
```

### Applying Set Operations

We now apply common set operations to compare observations from both days:
- `union()` – all unique trees from both days
- `intersection()` – trees seen on both days
- `difference()` – trees seen on one day but not the other
- `symmetric_difference()` – trees seen on only one of the days


```python
# Union of both sets
union_result = day_1_trees.union(day_2_trees)

# Trees seen on both days
both_days = day_1_trees.intersection(day_2_trees)

# Trees seen on day 2 but not on day 1
day2_only = day_2_trees.difference(day_1_trees)

# Trees seen on only one of the two days
one_set_result = day_1_trees.symmetric_difference(day_2_trees)

union_result, both_days, day2_only, one_set_result
```

### Subset Testing

We check whether all trees seen on both days are contained within the day 1 observations.


```python
both_days.issubset(day_1_trees)
```

### Testing Set Membership

We now test whether all elements of a predefined set are contained within another set using a single command.


```python
hardwood_trees = {'Yellowwood', 'Leadwood', 'Blackwood'}

hardwood_trees.issubset(day_2_trees)
```
